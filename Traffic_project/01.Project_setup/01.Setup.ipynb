```sql
%python
dbutils.widgets.text(name="env",defaultValue ="", label="This is the environment value")
env = dbutils.widgets.get("env")
```

```sql
%python
def bronze_table(environment):
    print(f"""Creating bronze table in {environment}""")
    spark.sql(f"""CREATE TABLE IF NOT EXISTS traffic_data.bronze.raw_traffic
        (
                            Record_ID INT,
                            Count_point_id INT,
                            Direction_of_travel VARCHAR(255),
                            Year INT,
                            Count_date VARCHAR(255),
                            hour INT,
                            Region_id INT,
                            Region_name VARCHAR(255),
                            Local_authority_name VARCHAR(255),
                            Road_name VARCHAR(255),
                            Road_Category_ID INT,
                            Start_junction_road_name VARCHAR(255),
                            End_junction_road_name VARCHAR(255),
                            Latitude DOUBLE,
                            Longitude DOUBLE,
                            Link_length_km DOUBLE,
                            Pedal_cycles INT,
                            Two_wheeled_motor_vehicles INT,
                            Cars_and_taxis INT,
                            Buses_and_coaches INT,
                            LGV_Type INT,
                            HGV_Type INT,
                            EV_Car INT,
                            EV_Bike INT,
                            Extract_Time TIMESTAMP
        )"""
        )
```

```sql
%python
bronze_table(env)
```

```sql
%python
def createTable_bronze_raw_roads(environment):
    print(f"""Creating raw_roads bronze table in {environment}""")
    spark.sql(f"""CREATE Table if not exists traffic_data.bronze.raw_roads(
                            Road_ID INT,
                            Road_Category_Id INT,
                            Road_Category VARCHAR(255),
                            Region_ID INT,
                            Region_Name VARCHAR(255),
                            Total_Link_Length_Km DOUBLE,
                            Total_Link_Length_Miles DOUBLE,
                            All_Motor_Vehicles DOUBLE
    );""")
```

```sql
%python
createTable_bronze_raw_roads(env)
```
