```python
def replace_null(df):
    columns = df.schema.names
    print('Remove Null with Unknown for string')
    df1 = df.fillna('Unkown', columns )
    print('Remove Null with 0 for interger')
    df_clean = df1.fillna(0,columns)
    return df_clean
    
```
