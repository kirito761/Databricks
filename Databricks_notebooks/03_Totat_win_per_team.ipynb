```python
data = [
("India", "Australia", "India"),
("England", "Pakistan", "England"),
("South Africa", "New Zealand", "New Zealand"),
("West Indies", "Sri Lanka", "Sri Lanka"),
("India", "Pakistan", "India"),
("Australia", "England", "Australia"),
("New Zealand", "Sri Lanka", "New Zealand"),
("South Africa", "West Indies", "South Africa"),
("India", "England", "India"),
("Pakistan", "Australia", "Pakistan")
]

columns = ["Team_1", "Team_2", "Winner"]
```

```python
df = spark.createDataFrame(data,columns)
df.display()
```

```python
from pyspark.sql import functions as F
#from pyspark.sql import functions as F
```

```python
df1 =(
    df.select(F.col("Team_1").alias("Team"),"Winner")
        .union(df.select(F.col("Team_2").alias("Team"),"Winner"))
        .withColumn("Winner_Flag",F.when(F.col("team") == F.col("Winner"), 1).otherwise(0))
        .select("Team","Winner_Flag")
    )
df1.display()
```

```python
df2 = df1.groupBy("Team").agg(F.sum("Winner_Flag").alias("Total_win")).sort("Total_win",ascending=False)
df2.display() 
```
