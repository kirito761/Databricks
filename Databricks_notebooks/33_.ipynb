```python
from pyspark.sql.functions import *
```

```python
data=[(1,"Sagar-Prajapati"),(2,"Alex-John"),(3,"John Cena"),(4,"Kim Joe")]
schema="ID int,Name string"
df=spark.createDataFrame(data,schema)
df.display()
```

```python
df1 = df.withColumn("new_name",regexp_replace(df.Name,r"( )","-"))
df1.display()
```

```python
df2 = df.withColumn("new_name",regexp_extract(df.Name,r"([a-zA-z]+-[a-zA-Z]+)",1))
df2.display()
```

```python
df2.filter(df2.new_name != '').display()
```

```python
df3 = df2.filter(df2.new_name == '').count()
df3
```

```python
if(df3):
    raise Exception("check name input")
else:
    print("No exception")
```
