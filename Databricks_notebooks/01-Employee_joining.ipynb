```python
df = spark.table(f"dev.practice.emp01")
```

```python
from pyspark.sql.functions import concat_ws
df1 = df.withColumn("Name",concat_ws(" ","first_name","last_name"))

df1.display()
```

```python
from pyspark.sql.functions import col,when
df2 = df1.withColumn("offredSalary",(col("current_ctc") * when(col("department") == "Engineering",1.25).when(col("department") == "IT",1.17).otherwise(1)))
df2.display()
```

```python
df3 = df2.select("id","Name","expected_doj","department","current_ctc","offredSalary")
df3.display()
```
