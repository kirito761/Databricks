```python
df = spark.createDataFrame([(1234,), (3245,), (2435,), (13,)], ["id"])
df.display()
```

```python
from pyspark.sql import functions as F
```

```python
df1 = (df
       .withColumn("digits", F.explode(F.split(F.col("id").cast("string"),"")))
       .withColumn("digits", F.col("digits").cast("int"))
       .groupBy("Id")
       .agg(F.sum("digits"))
)

df1.display()
```
