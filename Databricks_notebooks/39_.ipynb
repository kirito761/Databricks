```python
from pyspark.sql import functions as F
```

```python
data=[('Genece' , 2 , 75000),
('𝗝𝗮𝗶𝗺𝗶𝗻' , 2 , 80000 ),
('𝗣𝗮𝗻𝗸𝗮𝗷' , 2 , 80000 ),
('Tarvares' , 2 , 70000),
('Marlania' , 4 , 70000),
('Briana' , 4 , 85000),
('𝗞𝗶𝗺𝗯𝗲𝗿𝗹𝗶' , 4 , 55000),
('𝗚𝗮𝗯𝗿𝗶𝗲𝗹𝗹𝗮' , 4 , 55000),  
('Lakken', 5, 60000),
('Latoynia' , 5 , 65000) ]
schema="emp_name string,dept_id int,salary int"
df=spark.createDataFrame(data,schema)
# df.display()
```

```python
df1 = df.groupBy("dept_id").agg(F.max("salary").alias("max_salary"), F.min("salary").alias("min_salary")).withColumnRenamed("dept_id", "dept_id1")
# display(df1)
# df.display()
df2 = df.join(F.broadcast(df1), df.dept_id == df1.dept_id1,"inner").drop("dept_id1").filter((df.salary == df1.max_salary) | (df.salary == df1.min_salary))
df2 = df2.groupBy("dept_id","salary").agg(F.collect_list("emp_name").alias("emp_names"))
display(df2)
```
