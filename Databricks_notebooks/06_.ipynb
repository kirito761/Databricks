```python
data = [
    ("John",  "Doe",    1111, "123 Main St"),
    ("Alice", "Smith",  2222, "456 Elm St"),
    ("Bob",   "Brown",  1111, "789 Pine Ave"),
    ("Eve",   "Davis",  3333, None),
    ("Mike",  "Wilson", 4444, "999 Oak St"),
    ("Sam",   "White",  1111, "222 Birch Rd"),
    ("Anna",  "Clark",  5555, None),
    ("Leo",   "Adams",  6666, "777 Cedar St")
]

columns = [
    "first_name",
    "last_name",
    "pin",
    "address"
]

df = spark.createDataFrame(data, columns)

df.display()
```

```python
from pyspark.sql import functions as F
```

```python
df1 = (
    df.withColumn("full_name", F.concat_ws(" ","first_name", "last_name"))
)
df1.display()
```

```python
df2 = (
    df1.withColumn("new_address", F.when(F.col("pin") == 1111, F.col("address")).otherwise(F.lit(None)))
)
df2.display()
```

```python
df3 = df2.withColumn("address", F.when(F.col("address").isNull(),"Unknown").otherwise(F.col("address"))).sort(F.col("pin"))
df3.display()
```
