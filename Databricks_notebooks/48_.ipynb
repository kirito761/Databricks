```python
data = [(101, 'IT', 1000), (102, 'HR', 900)]
columns = ["empid", "dept", "salary"]
df = spark.createDataFrame(data, columns)
df.display()
```

```python
df1 = df
for i in df1.columns:
    df1 = df1.withColumnRenamed(i, "pref_" + i)
df1.display()
```

```python
from pyspark.sql.functions import *
```

```python
df2 = df
df2 = df2.select([col(i).alias("pref_" + i) for i in df.columns])
df2.display()
```
