```python
from pyspark.sql import functions as F
from pyspark.sql import Window
```

```python
df1 = (
    df.withColumn("result", F.when((F.col("home_team") == F.col("result")), "W").otherwise("L"))
    .withColumnRenamed("home_team","team")
    .drop("away_team")
)

df1.display()
```

```python
df2 = (
    df.withColumn("result", F.when((F.col("away_team") == F.col("result")), "W").otherwise("L"))
    .withColumnRenamed("away_team","team")
    .drop("home_team")
)

df2.display()
```

```python
df_final = (
    df1.union(df2)
        #.orderBy("match_no")
        .orderBy("team","match_no")
        .withColumn("loss_group", F.sum(F.when(F.col("result") == "L", 1).otherwise(0)).over(Window.partitionBy("team").orderBy("match_no")))
        .withColumn("win_streak",
                    F.when(F.col("result") == "W", F.row_number().over(Window.partitionBy("team", "loss_group").orderBy("match_no")) -1)
                    .otherwise(0)
    ).orderBy("team", "match_no")
    .groupBy("team").agg(F.max("win_streak").alias("max_streak"))
        #.withColumn("streak", F.rank().over(Window.partitionBy("team","result").orderBy("match_no")))
        #.withColumn("streak2", F.rank().over(Window.partitionBy("team").orderBy("match_no")))
        #.groupBy("team","result")
        #.agg(F.count("result").alias("count"))
)
df_final.display()
```
