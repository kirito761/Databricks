```python
from pyspark.sql import functions as F
```

```python
data=[(1,5),(2,6),(3,5),(3,6),(1,6)]
schema="customer_id int,product_key int"
customer_df=spark.createDataFrame(data,schema)
customer_df.display()

data=[(5,),(6,)]
schema="product_key int"
product_df=spark.createDataFrame(data,schema)
product_df.display()
```

```python
customer_df = customer_df.groupBy("customer_id").agg(F.countDistinct("product_key").alias("cnt_product"))
display(customer_df)
```

```python
product_df = product_df.agg(F.countDistinct("product_key").alias("cnt_product"))
display(product_df)
```

```python
df = customer_df.join(product_df, product_df. cnt_product == customer_df.cnt_product, "inner").select("customer_id")
display(df)
```
