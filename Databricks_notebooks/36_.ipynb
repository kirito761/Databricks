```python
from pyspark.sql import functions as F
```

```python
data = [(1, "abc@gmail.com"), (2, "bcd@gmail.com"), (3, "abc@gmail.com")]
schema = "ID int,email string"
df = spark.createDataFrame(data, schema)
```

```python
df.display()
```

```python
df.groupBy("email").agg(F.collect_list("ID")).display()
```

```python
df.groupBy("email").count().filter(F.col("count")>1).select("email").display()
```

```python
df.createOrReplaceTempView("test")
```

```python
%sql
select email from (select count(*) as count,email from test group by email having count > 1);
```

```python
df1 = sql("""
         select email from (select count(*) as count,email from test group by email having count > 1);
         """)
df1.display()
```
