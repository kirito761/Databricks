```python
%sql
with employee_rank as (
    select e.Id,
    e.dep_id,
    e.name,
    e.salary,
    d.dep_name,
    dense_rank() over (partition by e.dep_id order by e.salary desc) as rank 
    from dev.practice.Employee e
    join dev.practice.department d on e.dep_id = d.id
)

select name,dep_name from employee_rank where rank = 2;
```
