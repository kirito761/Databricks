```python
from pyspark.sql import functions as F
```

```python
data = [("John,Smith", "Canada"), ("Mike,David", "USA")]
df = spark.createDataFrame(data,["Names","Country"])
```

```python
df.display()
```

```python
df= df.withColumn("Names",F.split(F.col("Names"),","))
df.display()
```

```python
df = df.withColumn("Names",F.explode(F.col("Names")))
df.display()
```
