```python
data = [
    (1, "Alice", 1500, "2023-05-15"),
    (2, "Bob", 500, "2023-02-20"),
    (2, "Bob", 700, "2023-04-22"),
    (3, "Charlie", 1200, "2022-12-10"),
    (4, "Donald", 1200, "2024-12-10"),
    (4, "Donald", 1400, "2024-11-10"),
    (5, "Tom", 800, "2023-12-08")
]

schema = ['id', 'name', 'purchase_amount', 'purchase_date']

df = spark.createDataFrame(data, schema)
df.display()
```

```python
from pyspark.sql import functions as F

df1 = (
    df.withColumn("year",F.year(F.col("purchase_date")))
        .groupBy("name","year").agg(F.sum("purchase_amount").alias("total_amount"))
        .filter("total_amount > 1000")
        .filter( F.col("year") == F.year(F.current_date())-2)
       )
df1.display()
```
