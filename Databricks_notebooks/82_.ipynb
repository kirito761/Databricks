```python
friends_data = [(1, 2),
(1, 3),
(1, 4),
(2, 1),
(3, 1),
(3, 4),
(4, 1),
(4, 3)]

friend_schema = "user_id int , friend_id int"
friends_df = spark.createDataFrame(data =friends_data , schema = friend_schema )

likes_data = [
(1, 'A'),
(1, 'B'),
(1, 'C'),
(2, 'A'),
(3, 'B'),
(3, 'C'),
(4, 'B')
]


like_schema = "user_id int , page_id string"
likes_df = spark.createDataFrame(data =likes_data , schema = like_schema )

display(friends_df)
display(likes_df)
```

```python
from pyspark.sql import functions as F
```

```python
friends_df = (
    friends_df
        .join(likes_df, "user_id","inner")
        .withColumn("combi", F.concat(F.col("friend_id"),F.col("page_id")))
)

display(friends_df)
```

```python
likes_df = (
    likes_df.withColumn("combi", F.concat(F.col("user_id"),F.col("page_id")))

)

display(likes_df)
```

```python
friend_liked =  [row.combi for row in likes_df.collect()]

print(friend_liked)
```

```python
friends_df = (
    friends_df.filter(~F.col("combi").isin(friend_liked))
    .select("friend_id","page_id")
    .orderBy("friend_id","page_id")
    .distinct()
)

friends_df.display()
```
