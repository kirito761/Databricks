```python
data = [
    ('Virat Kohli',),
    ('Narendra Damodardas Modi',),
    ('Salman',)
]

schema = "celebrity_name string"

df = spark.createDataFrame(data = data , schema = schema)
df.display()
```

```python
from pyspark.sql import functions as F
```

```python
df1 = (
    df.withColumn("split", F.split(F.col("celebrity_name")," "))
        .withColumn("firstname", F.col("split")[0])
        .withColumn("middle_name", F.when(F.size(F.col("split")) == 3, F.col("split")[1]).otherwise(None))
        .withColumn("lastname", F.when(F.size(F.col("split")) == 3, F.col("split")[2]).when(F.size(F.col("split")) == 2, F.col("split")[1]).otherwise(None))
)
df1.display()
```
