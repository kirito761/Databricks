```python
data = [
    (1,5),
    (1,3),
    (1,6),
    (2,1),
    (2,6),
    (3,9),
    (4,1),
    (7,2),
    (8,3)
]

schema ="user1 int, user2 int"

df = spark.createDataFrame(data = data , schema = schema)
df.show()
```

```python
from pyspark.sql import functions as F
```

```python
df1  = df.withColumnsRenamed({
    'user1' : 'friend1',
    "user2" : 'friend2'
})

df2  = df.withColumnsRenamed({
    'user1' : 'friend2',
    "user2" : 'friend1'
}).select("friend1","friend2")

df1.display()
df2.display()
```

```python
df_final = (
    df1.union(df2)
)

df_final.display()
```

```python
df3 = (
    df_final.groupBy("friend1").agg(F.count("friend2").alias("count"))
)

df3.display()
```

```python
df5 = df3.select(F.count('*').alias('Total_count'))
df5.display()

df4 = (
    df3.join(df5, how='inner')
    .withColumn("percentage",
                    F.round(F.expr("count/Total_count * 100"),3)
                    )
    .drop("count","Total_count")
)

df4.display()
```
