```python
df = spark.read.format("json").option("multiline","true").load("/Volumes/dev/practice/my_volume/56.json")
display(df)
```

```python
from pyspark.sql import functions as F
```

```python
df1 = (
    df.withColumn("new_batters",F.explode(F.col("batters.batter")))
      .withColumn("new_toppings",F.explode(F.col("topping")))  
      .select("id","name","ppu","type",
              F.col("new_batters.id").alias("batter_id")
              ,F.col("new_batters.type").alias("batter_type")
              ,F.col("new_toppings.id").alias("topping_id")
              ,F.col("new_toppings.type").alias("topping_type")
              )
)
display(df1)
```
