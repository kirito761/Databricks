```python
job_skills_data = [
    (1, 'Data Engineer', 'SQL'),
    (2, None, 'Python'),
    (3, None, 'AWS'),
    (4, None, 'Snowflake'),
    (5, None, 'Apache Spark'),
    (6, 'Web Developer', 'Java'),
    (7, None, 'HTML'),
    (8, None, 'CSS'),
    (9, 'Data Scientist', 'Python'),
    (10, None, 'Machine Learning'),
    (11, None, 'Deep Learning'),
    (12, None, 'Tableau')
]

job_skills_schema = "row_id int , job_role string , skills string"

job_skills_df = spark.createDataFrame(data = job_skills_data , schema = job_skills_schema)
job_skills_df.display()
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window
```

```python
df = (
    job_skills_df
        .withColumn("flag", F.when(F.col("job_role").isNotNull(),1).otherwise(0))
        .withColumn("grp", F.sum(F.col("flag")).over(Window.orderBy(F.col("row_id"))))
        .withColumn("job_role", F.first(F.col("job_role")).over(Window.partitionBy(F.col("grp")).orderBy("row_id")))
)

df.display()
```
