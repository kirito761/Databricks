```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
data=[('IT', 'M'), ('IT', 'F'), ('IT', 'M'), ('IT', 'M'), ('HR', 'F'), ('HR', 'F'), ('HR', 'F'), ('HR', 'F'), ('HR', 'F'), ('Sales', 'M'),
('Sales', 'M'), ('Sales', 'F'), ('Sales', 'M'), ('Sales', 'M'), ('Sales', 'F')]
schema=["DeptName", "Gender"]
df1=spark.createDataFrame(data, schema)
display(df1)
```

```python
df2 = df1.select("DeptName", when(col("Gender") == 'M',1).alias("Male"),when(col("Gender") == 'F',1).alias("Female"))
df2.display()
```

```python
df2 = df2.groupBy("DeptName").agg(sum("Male").alias("MaleEmp"),sum("Female").alias("FemaleEmp"))
df2.display()
```

```python
df1 = df1.groupBy("DeptName").agg(count("Gender").alias("Gender"))  
df1.display()
```

```python
df_join = df2.join(df1, df2.DeptName == df1.DeptName, "inner").drop(df1.DeptName)
df_join.display()
```

```python
df_join = df_join.fillna(
    {
        "MaleEmp": 0,
        "FemaleEmp": 0,
        "Gender": 0
    }
)

df_join.display()
```
