```python
data = [
    (1, "A", 20, "31|32|34"),
    (2, "B", 21 ,"21|32|43"),
    (3, "C", 22, "21|32|11"),
    (4, "D", 23 , "10|12|12")
    
]

data_schema=['ID','Name','Age', 'Marks']

df = spark.createDataFrame(data,data_schema)
df.display()
```

```python
from pyspark.sql import functions as F
```

```python
(
    df.withColumn('A',F.split(F.col("Marks"),"\\|")[0])
    .withColumn('B',F.split(F.col("Marks"),"\\|")[1])
    .withColumn('C',F.split(F.col("Marks"),"\\|")[2])
 .display()
)
```
