```python
data = ["1-A-12-2-B-23-3-C-34-4-D-14"]
schema = ["value"]
df= spark.createDataFrame(data,schema)
df.show()
```

```python
from pyspark.sql.functions import *
```

```python
df1 = df.withColumn("newValue", regexp_replace("value","(.*?\\-){3}","$0,"))
df1.display()
```

```python
df2 = df1.withColumn("newValue1",split(col("newValue"),"\\-,"))
df2.display()
```

```python
df3 = df2.withColumn("value2",explode(col("newValue1")))
df3.display()
```

```python
df3.select("value2").display()
```
