```python
data = [
    (1, "A", 23),
    (2, "B", None),
    (3, "C", 56),
    (4, None, None),
    (5, None, None)
]

data_schema=['ID','Name','Age']

df=spark.createDataFrame(data,data_schema)
```

```python
from pyspark.sql import functions as F
df1 = df.select([F.count(F.when(F.col(i).isNull(),1)).alias("count " + i) for i in df.columns])
df1.display()
```
