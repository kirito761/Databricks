```python
from pyspark.sql.functions import explode
```

```python
sample=[(1,["munna","singh"])
        ,(2,["Saurabh", "singh"])
        ,(3,["Dev", "Billu"])
        ,(4,["Sumit"])
        ]
column= ["id", "name"]        

df1= spark.createDataFrame(data=sample, schema=column)
df1.display()
```

```python
df1.withColumn("Name",explode(df1.name)).display()
```
