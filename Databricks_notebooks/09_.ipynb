```python
data = [
(1, 'Alice', ["Python", "SQL"]),
(2, "Bob", ["Java", "ADF"]),
(3, 'Charlie', ["Scala", "PySpark", "Databricks"])
]
# Columns
columns = ['id', 'name','skills']

df = spark.createDataFrame(data, columns)
display(df)
```

```python
from pyspark.sql.functions import explode
df.select("id","name",explode("skills").alias("skill")).display()
```
