```python
data=[('John Doe','{"street": "123 Main St", "city": "Anytown"}'),('Jane Smith','{"street": "456 Elm St", "city": "Othertown"}')]
df=spark.createDataFrame(data,schema="name string,address string")
display(df)
```

```python
from pyspark.sql.functions import *
```

```python
df1 = df.withColumn("address1",from_json("address","street string,city string"))
df1.display()
```

```python
df1.select("*",col("address1").street.alias("street"),col("address1").city.alias("city")).drop("address","address1").display()
```

```python
df1.select(df1.name,df1.address1.street.alias("street"), df1.address1.city.alias("city")).display()
```
