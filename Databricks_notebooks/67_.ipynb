```python
from pyspark.sql import SparkSession
from pyspark.sql import Row

# Create data
data = [
    (111, 'Apple'),
    (111, 'Banana'),
    (112, 'Apple'),
    (113, 'Apple'),
    (113, 'Orange'),
    (114, 'Apple'),
    (114, 'Banana'),
    (114, 'Orange')
]

# Create DataFrame
product_df = spark.createDataFrame(data, ["User_ID", "Prod_Name"])
product_df.display()
```

```python
from pyspark.sql import functions as F
(
    product_df
    .groupBy("User_ID").agg(F.collect_set("Prod_Name").alias("Products"))
    .filter(F.sort_array(F.col("Products")) == F.array(F.lit("Apple"),F.lit("Banana")))
    .select("User_ID")
    .display()
)
```
