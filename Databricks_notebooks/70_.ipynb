```python
from pyspark.sql import functions as F
```

```python
data = [
    (1, 'X'),
    (2, 'Y'),
    (3, 'Z')
]

target = [
    (1, 'X'),
    (2, 'Y'),
    (4,'N'),
    (5,'O')
]
```

```python
df_source = spark.createDataFrame(data, ['id', 'value_source'])
df_target = spark.createDataFrame(target, ['id', 'value_target'])
display(df_source)
display(df_target)
```

```python
df_source1 = df_source.join(df_target,"id","left_anti").withColumn("cooment", F.lit("only in source"))
df_source1.display()
```

```python
df_target1 = df_target.join(df_source,"id","left_anti").withColumn("cooment", F.lit("only in target"))
df_target1.display()
```

```python
df_result = df_source1.union(df_target1).drop("value_source")
df_result.display()
```
