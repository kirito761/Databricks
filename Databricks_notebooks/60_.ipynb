```python
data="""Dept,Employee,Salary
HR,Alice,50000
HR,Bob,55000
HR,Charlie,52000
Finance,David,75000
Finance,Eve,70000
Finance,Frank,78000
IT,Grace,60000
IT,Hank,62000
IT,Ivy,61000"""

dbutils.fs.put('/Volumes/dev/practice/my_volume/60.csv',data,True)
```

```python
df = spark.read.format('csv').option("header",True).option("inferSchema",True).load('/Volumes/dev/practice/my_volume/60.csv')
display(df)
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import *
```

```python
df2 = (
    df.withColumn("rank",
                  F.rank().over(
                      Window.partitionBy("Dept").orderBy(F.col("Salary").desc())
                  ))
)
df2.display()
```

```python
(
    df.groupBy("Dept")
    .agg(
        F.max("Salary").alias("max"),
        F.min("Salary").alias("min")
    )
    .display()
)
```
