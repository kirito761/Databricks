```python
data = [
    (1, "Sagar", 23, "Male", 68.0),
    (2, "Kim", 35, "Female", 90.2),
    (3, "Alex", 40, "Male", 79.1),
]
schema = "Id int,Name string,Age int,Gender string,Marks float"
df = spark.createDataFrame(data, schema)
df.display()
```

```python
df.dtypes
```

```python
for i in df.dtypes:
    if(i[1]=='int'):
        print(i[0])
```

```python
df_type_list= set(i[1] for i in df.dtypes)
df_type_list
```

```python
%sql
CREATE VOLUME IF NOT EXISTS dev.practice.my_volume;
```

```python
dbutils.fs.mkdirs("dbfs:/Volumes/dev/practice/my_volume/output/")
```

```python
for i in df_type_list:
    col = []
    for j in df.dtypes:
        if(j[1]==i):
            col.append(j[0])
    print(col)
    df.select(col).write.mode('overwrite').save(f'/Volumes/dev/practice/my_volume/output/{i}')
```
