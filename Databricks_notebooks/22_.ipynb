```python
from pyspark.sql.types import *
schema = StructType([
StructField("ActorId", IntegerType (), True),
StructField("DirectorId", IntegerType(), True),
StructField("timestamp", IntegerType(), True)
])

data = [
(1, 1, 0),
(1, 1, 1),
(1, 1, 2),
(1, 2, 3),
(1, 2, 4),
(2, 1, 5),
(2, 1, 6)
]
df = spark.createDataFrame(data, schema)
display(df)
```

```python
df = df.sort("ActorId","DirectorId")
df.display()
```

```python
df.groupBy("ActorId","DirectorId").count().alias("count").filter("count>2").drop("count").display()
```
