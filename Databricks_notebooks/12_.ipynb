```python
data=[('Goa', '', 'AP'),('', 'AP', None), (None, '', 'Bglr')]
columns=["city1", "city2", "city3"]
df=spark.createDataFrame(data, columns)
display(df)
```

```python
from pyspark.sql import functions as F
df.withColumn("first_city", F.coalesce(
    F.when(df.city1 == "",None).otherwise(df.city1),
    F.when(df.city2 == "",None).otherwise(df.city2),
    F.when(df.city3 == "",None).otherwise(df.city3)
)).select("first_city").display()
```
