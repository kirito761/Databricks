```python
from pyspark.sql import functions as F
```

```python
data = [
    ("john", "tomato", 2),
    ("𝚋𝚒𝚕𝚕", "𝚊𝚙𝚙𝚕𝚎", 2),
    ("john", "𝚋𝚊𝚗𝚊𝚗𝚊", 2),
    ("john", "tomato", 3),
    ("𝚋𝚒𝚕𝚕", "𝚝𝚊𝚌𝚘", 2),
    ("𝚋𝚒𝚕𝚕", "𝚊𝚙𝚙𝚕𝚎", 2),
]
schema = "name string,item string,weight int"
df = spark.createDataFrame(data, schema)
df.display()
```

```python
df1 = df.groupBy("name","item").agg(F.sum(F.col("weight")).alias("weight"))
df1.display()
```

```python
df2 = df1.groupBy("name").agg(F.collect_list(F.struct("item","weight")).alias("list"))
df2.display()
```

```python
df3 = df2.withColumn("list",F.concat(F.lit("("),F.col("list")[0]["item"],F.lit(","),F.col("list")[0]["weight"],F.lit(")"),F.lit(","),F.lit("("),F.col("list")[1]["item"],F.lit(","),F.col("list")[1]["weight"],F.lit(")") ))
df3.display()
```
