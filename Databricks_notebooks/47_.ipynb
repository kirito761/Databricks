```python
data = [(1, None, 'ab'),
    (2, 10, None),
    (None, None, 'cd')]
columns = ['col1', 'col2', 'col3']
df = spark.createDataFrame(data, columns)
df.display()
```

```python
from pyspark.sql.functions import *
```

```python
df1 = df.select([sum(col(i).isNull().cast("int")).alias(i) for i in df.columns])
display(df1)
```
