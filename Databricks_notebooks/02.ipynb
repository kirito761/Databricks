```python
data = [
    (101, "Alice", "Math", 85),
    (101, "Alice", "Chemistry", 85),
    (101, "Alice", "Physics", 78),

    (102, "Bob", "Math", 90),
    (102, "Bob", "Physics", 88),

    (103, "Charlie", "Math", 75),
    (103, "Charlie", "Chemistry", 75),

    (104, "David", "Math", 88),
    (104, "David", "Chemistry", 88),
    (104, "David", "Physics", 95),

    (105, "Eve", "Chemistry", 91),
]

columns = ["student_id", "student_name", "subject", "marks"]
df = spark.createDataFrame(data, columns)
```

```python
df.display()
```

```python
from pyspark.sql import functions as F
```

```python
df1 = df.filter(F.col("subject").isin("Math","Chemistry"))
df1.display()
```

```python
from pyspark.sql.window import  Window
```

```python
window = Window.partitionBy("student_id","student_name").orderBy("subject")
df2 = df1.withColumn("prevMarks",F.lag("marks").over(window))
df2.display()
```

```python
df3 = df2.filter(F.col("marks") == F.col("prevMarks")).select("student_id","student_name")
df3.display()
```
