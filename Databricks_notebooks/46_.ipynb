```python
data1 = [(1, 'Bob'), (2, 'Alice'), (3, 'Tom')]
data2 = [(1, 'Bob'), (3, 'Tom')]

df1 = spark.createDataFrame(data1, ["id", "name"])
df2 = spark.createDataFrame(data2, ["id", "name"])

df1.display()
df2.display()
```

```python
df1.subtract(df2).display()
```

```python
df3 = df1.join(df2,"id","left_anti")
display(df3)
```
