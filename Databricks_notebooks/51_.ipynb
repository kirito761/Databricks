```python
from pyspark.sql.functions import col,min,max
data = [
 (1, ),
 (2,),
 (3,),
 (6,),
 (7,),
 (8,)]
schema="Id int"
df = spark.createDataFrame(data,schema=schema)
df.display()
```

```python
df1 = df.select(max(col('Id').alias("max")),min(col('Id').alias("min")))
df1.display()
df2 = spark.range(df1.first()[1],df1.first()[0]+1)
df2.display()
```

```python
df3 = df2.subtract(df)
display(df3)
```
