```python
from pyspark.sql import functions as F
```

```python
data = [("Raj","Doe",None),
 (None,"Samuel","VIZAG"),
 ("David","Smith", None),
 ("Samson",None, "HYD"),
 ("Immi", "Steve", "BNG"),
 (None, None, None)]

columns = ["Firstname", "Lastname", "City"]

df = spark.createDataFrame(data,columns)
df.display()
```

```python
for i in df.columns:
    total = df.select(F.col(i)).count()
    null = df.filter(F.col(i).isNull()).count()
    percentage = (null/total)*100
    print(i, " ",percentage)
```
