```python
from pyspark.sql import functions as F
data = [
    ("Pepperoni", 0.50),
    ("Sausage", 0.70),
    ("Chicken", 0.55),
    ("Extra Cheese", 0.40),
]

toppings = spark.createDataFrame(data, ["topping_name", "cost"])

toppings.display()

t1 = toppings.alias("t1")
t2 = toppings.alias("t2")
t3 = toppings.alias("t3")

df = t1.crossJoin(t2).crossJoin(t3).filter(F.expr("t1.topping_name < t2.topping_name") & F.expr("t2.topping_name < t3.topping_name"))

df.display()
```

```python
df1 = (df.withColumn("Ingridents",F.concat_ws(",",F.col("t1.topping_name"), F.col("t2.topping_name"),F.col("t3.topping_name")))
       .withColumn("total_cost", F.round(F.expr("t1.cost + t2.cost + t3.cost"),2))
       .select("Ingridents", "total_cost")
       .orderBy((F.col("total_cost")).desc(),
                F.col("Ingridents")
                )
       )

df1.display()
```
